```python
import cv2
import matplotlib.pyplot as plt
 
group_photo = cv2.imread('pic.jpg')
 
if group_photo is not None:
    group_photo = cv2.cvtColor(group_photo, cv2.COLOR_BGR2RGB)
    plt.imshow(group_photo)
    plt.axis('off')
    plt.show()
else:
    print("Failed to load image.")
```
